### B1. Valores faltantes (0.4 pts)
a) Construir la tabla con el **número** y el **porcentaje** de valores faltantes por columna, igual que en la demo.

*💡 Recordatorio: `df.isna().sum()` cuenta los faltantes por columna; `len(df)` da el total de filas; `pd.DataFrame({...})` arma la tabla y `.sort_values(...)` la ordena.*

```python
# ESCRIBIR EL CÓDIGO AQUÍ
falt = df.isna().sum()
falt_pct = (falt / len(df) * 100).round(1)
tabla_falt = pd.DataFrame({'n_faltantes': falt, '%': falt_pct}).sort_values('n_faltantes', ascending=False)
tabla_falt[tabla_falt['n_faltantes'] > 0]
```

### B2. El faltante escondido en `cargo_total` (0.4 pts)

En la demo vimos que `cargo_total` es de tipo texto porque contiene espacios en blanco `" "`.

a) Contar cuántas filas tienen `cargo_total` en blanco.
b) Mostrar el valor de `meses_antiguedad` de esos clientes.

*💡 Recordatorio: `pd.to_numeric(df["cargo_total"], errors="coerce")` convierte a número y deja `NaN` donde no se puede; `.isna()` marca esas filas y `.sum()` las cuenta. Después, `df.loc[filas, "meses_antiguedad"].value_counts()` muestra los valores de esas filas.*

```python
# ESCRIBIR EL CÓDIGO AQUÍ
# Convertir a numérico y detectar espacios en blanco
ct_num = pd.to_numeric(df["cargo_total"], errors="coerce")
mask_blanco = ct_num.isna()

# Contar filas
print("Filas con cargo_total en blanco:", mask_blanco.sum())

# Ver meses de antigüedad de esas filas
df.loc[mask_blanco, "meses_antiguedad"].value_counts()
```

c) ¿Qué tienen en común esos clientes? ¿Qué valor **debería** tener `cargo_total` en esos casos?

**Respuesta:**

_Todos esos clientes tienen 0 meses de antigüedad (recién dados de alta). En esos casos, `cargo_total` **debería ser 0.00** quetzales._

_Al revisar esos 61 registros, todos tienen `meses_antiguedad = 0`, es decir, son clientes recién dados de alta. Como aún no llevan ningún mes con la empresa ni han acumulado pagos, lo lógico es que su `cargo_total` sea 0.00 quetzales._

_**Por lo tanto, ese espacio en blanco no es un dato perdido, sino que simplemente debería ser cero.**_

En lugar de estudiar a 1534 clientes únicos, estaríamos inflando los números y podríamos sacar conclusiones equivocadas._
